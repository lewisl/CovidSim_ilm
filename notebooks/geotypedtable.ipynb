```julia
function buildgeodata(filename)
    tmp = Table(CSV.File(filename))
    
    Table(tmp, 
        density_factor = shifter(tmp.density,0.9,1.25), 
        anchor        = cs.quickdate(tmp.anchor),
        limit         = cs.quickdate(tmp.limit)
        )
end
```

```julia
geodf = buildgeodata(filename)
```
